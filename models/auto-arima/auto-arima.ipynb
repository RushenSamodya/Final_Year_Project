```python
# Auto-ARIMA Model for Kelani River Water Level Data
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import pmdarima as pm
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
import os
import warnings
warnings.filterwarnings('ignore')

# Set plotting style
plt.style.use('seaborn-v0_8')
sns.set(rc={'figure.figsize':(15, 8)})

def fit_auto_arima_model(train, test, seasonal=True, m=7):
    """
    Fit Auto-ARIMA model and make forecasts
    """
    print("Finding optimal ARIMA parameters...")
    
    # Create and fit model
    model = pm.auto_arima(
        train,
        start_p=0, start_q=0,
        max_p=5, max_q=5, max_d=2,
        m=m if seasonal else 1,  # seasonality
        seasonal=seasonal,
        stepwise=True,
        suppress_warnings=True,
        error_action="ignore",
        trace=True
    )
    
    print(f"Best model: ARIMA{model.order}")
    if seasonal:
        print(f"Seasonal order: {model.seasonal_order}")
    
    # Forecast
    forecast = model.predict(n_periods=len(test))
    forecast_series = pd.Series(forecast, index=test.index)
    
    # Plot results
    plt.figure(figsize=(15, 6))
    plt.plot(train[-30:], label='Training Data (last 30 points)')
    plt.plot(test, label='Actual')
    plt.plot(forecast_series, label='Forecast')
    if seasonal:
        plt.title(f'Auto-ARIMA{model.order}{model.seasonal_order} - Forecast vs Actual')
    else:
        plt.title(f'Auto-ARIMA{model.order} - Forecast vs Actual')
    plt.legend()
    plt.grid(True)
    plt.tight_layout()
    plt.show()
    
    # Make sure both test and forecast don't have NaN values
    valid_idx = ~(np.isnan(test) | np.isnan(forecast))
    
    if sum(valid_idx) < 2:
        print("WARNING: Not enough valid data points for evaluation.")
        return model, forecast_series, {'rmse': np.nan, 'mae': np.nan, 'r2': np.nan}
    
    # Filter out NaN values
    test_clean = test[valid_idx]
    forecast_clean = forecast[valid_idx]
    
    # Calculate metrics
    try:
        rmse = np.sqrt(mean_squared_error(test_clean, forecast_clean))
        mae = mean_absolute_error(test_clean, forecast_clean)
        r2 = r2_score(test_clean, forecast_clean)
        
        print(f'RMSE: {rmse:.4f}')
        print(f'MAE: {mae:.4f}')
        print(f'R²: {r2:.4f}')
        
        return model, forecast_series, {'rmse': rmse, 'mae': mae, 'r2': r2}
    except Exception as e:
        print(f"Error calculating metrics: {str(e)}")
        return model, forecast_series, {'rmse': np.nan, 'mae': np.nan, 'r2': np.nan}

def find_best_auto_arima_model(train, test):
    """
    Try different Auto-ARIMA configurations and return the best one
    """
    models = [
        {"name": "Non-Seasonal ARIMA", "seasonal": False, "m": None},
        {"name": "Weekly Seasonal ARIMA", "seasonal": True, "m": 7},
        {"name": "Monthly Seasonal ARIMA", "seasonal": True, "m": 30}
    ]
    
    best_model = None
    best_forecast = None
    best_metrics = {"rmse": float('inf')}
    best_config = None
    
    for config in models:
        print(f"\nTrying {config['name']}...")
        try:
            m_value = config['m'] if config['seasonal'] else 1
            
            model, forecast_series, metrics = fit_auto_arima_model(
                train, 
                test, 
                seasonal=config['seasonal'],
                m=m_value
            )
            
            print(f"{config['name']} - RMSE: {metrics['rmse']:.4f}, MAE: {metrics['mae']:.4f}, R²: {metrics['r2']:.4f}")
            
            # Keep track of best model
            if not np.isnan(metrics['rmse']) and metrics['rmse'] < best_metrics['rmse']:
                best_model = model
                best_forecast = forecast_series
                best_metrics = metrics
                best_config = config
        except Exception as e:
            print(f"Error fitting {config['name']}: {str(e)}")
    
    if best_model is not None:
        print(f"\nBest Auto-ARIMA Model: {best_config['name']}")
        print(f"Order: {best_model.order}")
        if best_config['seasonal']:
            print(f"Seasonal Order: {best_model.seasonal_order}")
        print(f"RMSE: {best_metrics['rmse']:.4f}")
        print(f"MAE: {best_metrics['mae']:.4f}")
        print(f"R²: {best_metrics['r2']:.4f}")
    
    return best_config['name'], best_model, best_forecast, best_metrics

def analyze_water_level_station_auto_arima(file_path, station_name):
    """
    Process, analyze and build Auto-ARIMA model for water level station data
    """
    print(f"\n{'='*80}")
    print(f"Analyzing station: {station_name}")
    print(f"{'='*80}")
    
    # Read data
    try:
        df = pd.read_csv(file_path)
    except Exception as e:
        print(f"Error reading file {file_path}: {str(e)}")
        return None
    
    # Check if data exists
    if df.empty:
        print(f"No data found for {station_name}")
        return None
    
    # Parse dates and set as index
    try:
        df.columns = ['datetime', 'unit', 'water_level', 'label']
        df['datetime'] = pd.to_datetime(df['datetime'])
        df.set_index('datetime', inplace=True)
    except Exception as e:
        print(f"Error parsing dates: {str(e)}")
        print("Columns in file:", df.columns)
        return None
    
    # Display basic info
    print("\nData Overview:")
    print(f"Time Range: {df.index.min()} to {df.index.max()}")
    print(f"Number of records: {len(df)}")
    print(f"Number of missing values: {df['water_level'].isna().sum()}")
    
    # Handle missing values if any
    if df['water_level'].isna().any():
        df['water_level'] = df['water_level'].interpolate(method='time')
    
    # Plot water level data
    plt.figure(figsize=(15, 6))
    plt.plot(df['water_level'])
    plt.title(f'Water Level - {station_name}')
    plt.xlabel('Date')
    plt.ylabel('Water Level (m)')
    plt.grid(True)
    plt.tight_layout()
    plt.show()
    
    # Resample to daily frequency (using mean) to reduce noise
    daily_data = df['water_level'].resample('D').mean()
    
    # Handle any remaining NaN values in daily data
    if daily_data.isna().any():
        print(f"Filling {daily_data.isna().sum()} NaN values in daily data")
        daily_data = daily_data.interpolate(method='time')
    
    # Split data into train and test sets
    train_size = int(len(daily_data) * 0.8)
    train, test = daily_data[:train_size], daily_data[train_size:]
    
    print(f"\nTraining set size: {len(train)}")
    print(f"Test set size: {len(test)}")
    
    # Find best Auto-ARIMA model
    print("\nFinding best Auto-ARIMA model...")
    model_name, model, forecast, metrics = find_best_auto_arima_model(train, test)
    
    return {
        'station_name': station_name,
        'model_name': model_name,
        'model': model,
        'forecast': forecast,
        'metrics': metrics
    }

# Main execution
def main():
    base_dir = r"c:\Users\RushenSamodya\Documents\GitHub\Final_Year_Project\labeled_data"
    
    # Path to Kelani River water level data
    kelani_dir = os.path.join(base_dir, "water_level_data", "kelani_ganga")
    
    # Check if directory exists
    if not os.path.exists(kelani_dir):
        print(f"Directory does not exist: {kelani_dir}")
        return
    
    # Get all stations for Kelani River
    stations = [f for f in os.listdir(kelani_dir) if f.endswith('.csv')]
    
    if not stations:
        print(f"No station data found in {kelani_dir}")
        return
    
    print(f"Found {len(stations)} stations for Kelani River")
    
    # Results container
    results = []
    
    # Process each station
    for station_file in stations:
        station_path = os.path.join(kelani_dir, station_file)
        station_name = station_file.replace('.csv', '')
        
        # Analyze station data
        result = analyze_water_level_station_auto_arima(station_path, station_name)
        
        if result:
            results.append(result)
    
    # Compare results across stations
    if results:
        print("\n\nComparison of model performance across stations:")
        comparison = pd.DataFrame([
            {
                'Station': r['station_name'],
                'Best Auto-ARIMA Model': r['model_name'],
                'Order': str(r['model'].order),
                'RMSE': r['metrics']['rmse'],
                'MAE': r['metrics']['mae'],
                'R²': r['metrics']['r2']
            }
            for r in results
        ])
        
        print(comparison)
        
        # Save results to CSV
        comparison.to_csv('auto_arima_model_results.csv', index=False)
        print("Results saved to auto_arima_model_results.csv")

# Execute main function
if __name__ == "__main__":
    main()
```
