```python
# STL Decomposition + Forecasting for Kelani River Water Level Data
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from statsmodels.tsa.seasonal import STL
from statsmodels.tsa.forecasting.stl import STLForecast
from statsmodels.tsa.arima.model import ARIMA
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
import os
import warnings
warnings.filterwarnings('ignore')

# Set plotting style
plt.style.use('seaborn-v0_8')
sns.set(rc={'figure.figsize':(15, 8)})

def plot_stl_decomposition(result):
    """
    Plot STL decomposition components
    """
    fig = plt.figure(figsize=(15, 10))
    result.plot()
    plt.tight_layout()
    plt.show()

def fit_stl_forecast_model(train, test, period=7, seasonal=7, arima_order=(1, 1, 1)):
    """
    Fit STL decomposition with ARIMA forecasting
    """
    # Create and fit model
    stlf = STLForecast(
        train,
        ARIMA,
        model_kwargs={"order": arima_order},
        period=period,
        seasonal=seasonal
    )
    stlf_res = stlf.fit()
    
    # Plot decomposition
    plot_stl_decomposition(stlf_res)
    
    # Forecast
    forecast = stlf_res.forecast(len(test))
    forecast_series = pd.Series(forecast, index=test.index)
    
    # Plot results
    plt.figure(figsize=(15, 6))
    plt.plot(train[-30:], label='Training Data (last 30 points)')
    plt.plot(test, label='Actual')
    plt.plot(forecast_series, label='Forecast')
    plt.title(f'STL+ARIMA{arima_order} (period={period}, seasonal={seasonal}) - Forecast vs Actual')
    plt.legend()
    plt.grid(True)
    plt.tight_layout()
    plt.show()
    
    # Make sure both test and forecast don't have NaN values
    valid_idx = ~(np.isnan(test) | np.isnan(forecast))
    
    if sum(valid_idx) < 2:
        print("WARNING: Not enough valid data points for evaluation.")
        return stlf_res, forecast_series, {'rmse': np.nan, 'mae': np.nan, 'r2': np.nan}
    
    # Filter out NaN values
    test_clean = test[valid_idx]
    forecast_clean = forecast[valid_idx]
    
    # Calculate metrics
    try:
        rmse = np.sqrt(mean_squared_error(test_clean, forecast_clean))
        mae = mean_absolute_error(test_clean, forecast_clean)
        r2 = r2_score(test_clean, forecast_clean)
        
        print(f'RMSE: {rmse:.4f}')
        print(f'MAE: {mae:.4f}')
        print(f'R²: {r2:.4f}')
        
        return stlf_res, forecast_series, {'rmse': rmse, 'mae': mae, 'r2': r2}
    except Exception as e:
        print(f"Error calculating metrics: {str(e)}")
        return stlf_res, forecast_series, {'rmse': np.nan, 'mae': np.nan, 'r2': np.nan}

def find_best_stl_forecast_model(train, test):
    """
    Try different STL+ARIMA configurations and return the best one
    """
    configs = [
        {"name": "Weekly", "period": 7, "seasonal": 7, "arima_order": (1, 1, 1)},
        {"name": "Weekly Enhanced", "period": 7, "seasonal": 7, "arima_order": (2, 1, 2)},
        {"name": "Bi-weekly", "period": 14, "seasonal": 14, "arima_order": (1, 1, 1)},
        {"name": "Monthly", "period": 30, "seasonal": 30, "arima_order": (1, 1, 1)},
    ]
    
    best_model = None
    best_forecast = None
    best_metrics = {"rmse": float('inf')}
    best_config = None
    
    for config in configs:
        print(f"\nTrying {config['name']} STL+ARIMA model...")
        try:
            model, forecast_series, metrics = fit_stl_forecast_model(
                train, 
                test, 
                period=config['period'],
                seasonal=config['seasonal'],
                arima_order=config['arima_order']
            )
            
            print(f"{config['name']} - RMSE: {metrics['rmse']:.4f}, MAE: {metrics['mae']:.4f}, R²: {metrics['r2']:.4f}")
            
            # Keep track of best model
            if not np.isnan(metrics['rmse']) and metrics['rmse'] < best_metrics['rmse']:
                best_model = model
                best_forecast = forecast_series
                best_metrics = metrics
                best_config = config
        except Exception as e:
            print(f"Error fitting {config['name']}: {str(e)}")
    
    # Check if we found any valid model
    if best_model is not None:
        print(f"\nBest STL+ARIMA Model: {best_config['name']}")
        print(f"Parameters: period={best_config['period']}, seasonal={best_config['seasonal']}, ARIMA{best_config['arima_order']}")
        print(f"RMSE: {best_metrics['rmse']:.4f}")
        print(f"MAE: {best_metrics['mae']:.4f}")
        print(f"R²: {best_metrics['r2']:.4f}")
        
        return best_config['name'], best_model, best_forecast, best_metrics, best_config
    else:
        # If no valid model was found, use the first config as fallback
        print("\nWARNING: No valid model could be fitted. Using fallback configuration.")
        fallback_config = configs[0]  # Use the first config as fallback
        print(f"Using fallback configuration: {fallback_config['name']}")
        
        # Try one more time with fallback config
        try:
            # Attempt with fallback configuration
            fallback_model, fallback_forecast, fallback_metrics = fit_stl_forecast_model(
                train, 
                test, 
                period=fallback_config['period'],
                seasonal=fallback_config['seasonal'],
                arima_order=fallback_config['arima_order']
            )
            return fallback_config['name'], fallback_model, fallback_forecast, fallback_metrics, fallback_config
        except Exception as e:
            print(f"Fallback model also failed: {str(e)}")
            # Return dummy values if everything fails
            empty_forecast = pd.Series(np.nan, index=test.index)
            empty_metrics = {'rmse': np.nan, 'mae': np.nan, 'r2': np.nan}
            return "Failed", None, empty_forecast, empty_metrics, fallback_config

def analyze_water_level_station_stl(file_path, station_name):
    """
    Process, analyze and build STL+ARIMA model for water level station data
    """
    print(f"\n{'='*80}")
    print(f"Analyzing station: {station_name}")
    print(f"{'='*80}")
    
    # Read data
    try:
        df = pd.read_csv(file_path)
    except Exception as e:
        print(f"Error reading file {file_path}: {str(e)}")
        return None
    
    # Check if data exists
    if df.empty:
        print(f"No data found for {station_name}")
        return None
    
    # Parse dates and set as index
    try:
        df.columns = ['datetime', 'unit', 'water_level', 'label']
        df['datetime'] = pd.to_datetime(df['datetime'])
        df.set_index('datetime', inplace=True)
    except Exception as e:
        print(f"Error parsing dates: {str(e)}")
        print("Columns in file:", df.columns)
        return None
    
    # Display basic info
    print("\nData Overview:")
    print(f"Time Range: {df.index.min()} to {df.index.max()}")
    print(f"Number of records: {len(df)}")
    print(f"Number of missing values: {df['water_level'].isna().sum()}")
    
    # Handle missing values if any
    if df['water_level'].isna().any():
        df['water_level'] = df['water_level'].interpolate(method='time')
    
    # Plot water level data
    plt.figure(figsize=(15, 6))
    plt.plot(df['water_level'])
    plt.title(f'Water Level - {station_name}')
    plt.xlabel('Date')
    plt.ylabel('Water Level (m)')
    plt.grid(True)
    plt.tight_layout()
    plt.show()
    
    # Resample to daily frequency (using mean) to reduce noise
    daily_data = df['water_level'].resample('D').mean()
    
    # Handle any remaining NaN values in daily data
    if daily_data.isna().any():
        print(f"Filling {daily_data.isna().sum()} NaN values in daily data")
        daily_data = daily_data.interpolate(method='time')
    
    # Basic STL decomposition visualization
    print("\nPerforming initial STL decomposition to visualize components...")
    try:
        stl = STL(daily_data, period=7, seasonal=7)
        result = stl.fit()
        fig = plt.figure(figsize=(15, 10))
        result.plot()
        plt.suptitle(f'STL Decomposition - {station_name}')
        plt.tight_layout()
        plt.show()
    except Exception as e:
        print(f"Warning: Initial STL decomposition failed: {str(e)}")
    
    # Split data into train and test sets
    train_size = int(len(daily_data) * 0.8)
    train, test = daily_data[:train_size], daily_data[train_size:]
    
    print(f"\nTraining set size: {len(train)}")
    print(f"Test set size: {len(test)}")
    
    # Find best STL+ARIMA model
    print("\nFinding best STL+ARIMA model...")
    model_name, model, forecast, metrics, config = find_best_stl_forecast_model(train, test)
    
    return {
        'station_name': station_name,
        'model_name': model_name,
        'model': model,
        'forecast': forecast,
        'metrics': metrics,
        'config': config
    }

# Main execution
def main():
    base_dir = r"c:\Users\RushenSamodya\Documents\GitHub\Final_Year_Project\labeled_data"
    
    # Path to Kelani River water level data
    kelani_dir = os.path.join(base_dir, "water_level_data", "kelani_ganga")
    
    # Check if directory exists
    if not os.path.exists(kelani_dir):
        print(f"Directory does not exist: {kelani_dir}")
        return
    
    # Get all stations for Kelani River
    stations = [f for f in os.listdir(kelani_dir) if f.endswith('.csv')]
    
    if not stations:
        print(f"No station data found in {kelani_dir}")
        return
    
    print(f"Found {len(stations)} stations for Kelani River")
    
    # Results container
    results = []
    
    # Process each station
    for station_file in stations:
        station_path = os.path.join(kelani_dir, station_file)
        station_name = station_file.replace('.csv', '')
        
        # Analyze station data
        result = analyze_water_level_station_stl(station_path, station_name)
        
        if result:
            results.append(result)
    
    # Compare results across stations
    if results:
        print("\n\nComparison of model performance across stations:")
        comparison = pd.DataFrame([
            {
                'Station': r['station_name'],
                'Best STL+ARIMA Model': r['model_name'],
                'Period': r['config']['period'] if 'period' in r['config'] else 'N/A',
                'ARIMA Order': str(r['config']['arima_order']) if 'arima_order' in r['config'] else 'N/A',
                'RMSE': r['metrics']['rmse'],
                'MAE': r['metrics']['mae'],
                'R²': r['metrics']['r2']
            }
            for r in results
        ])
        
        print(comparison)
        
        # Save results to CSV
        comparison.to_csv('stl_forecast_model_results.csv', index=False)
        print("Results saved to stl_forecast_model_results.csv")

# Execute main function
if __name__ == "__main__":
    main()
```
