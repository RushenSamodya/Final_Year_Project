```python
# ETS Model for Kelani River Water Level Data
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from statsmodels.tsa.holtwinters import ExponentialSmoothing
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
import os
import warnings
warnings.filterwarnings('ignore')

# Set plotting style
plt.style.use('seaborn-v0_8')
sns.set(rc={'figure.figsize':(15, 8)})

def fit_ets_model(train, test, seasonal_periods=7, trend_type='add', seasonal_type='add'):
    """
    Fit ETS model and make forecasts
    """
    # Create and fit model
    model = ExponentialSmoothing(
        train,
        trend=trend_type,
        seasonal=seasonal_type,
        seasonal_periods=seasonal_periods,
        use_boxcox=False,
        initialization_method="estimated"
    )
    
    model_fit = model.fit()
    
    # Forecast
    forecast = model_fit.forecast(len(test))
    forecast_series = pd.Series(forecast, index=test.index)
    
    # Plot results
    plt.figure(figsize=(15, 6))
    plt.plot(train[-30:], label='Training Data (last 30 points)')
    plt.plot(test, label='Actual')
    plt.plot(forecast_series, label='Forecast')
    plt.title(f'ETS ({trend_type}, {seasonal_type}) - Forecast vs Actual')
    plt.legend()
    plt.grid(True)
    plt.tight_layout()
    plt.show()
    
    # Make sure both test and forecast don't have NaN values
    valid_idx = ~(np.isnan(test) | np.isnan(forecast))
    
    if sum(valid_idx) < 2:
        print("WARNING: Not enough valid data points for evaluation.")
        return model_fit, forecast_series, {'rmse': np.nan, 'mae': np.nan, 'r2': np.nan}
    
    # Filter out NaN values
    test_clean = test[valid_idx]
    forecast_clean = forecast[valid_idx]
    
    # Calculate metrics
    try:
        rmse = np.sqrt(mean_squared_error(test_clean, forecast_clean))
        mae = mean_absolute_error(test_clean, forecast_clean)
        r2 = r2_score(test_clean, forecast_clean)
        
        print(f'RMSE: {rmse:.4f}')
        print(f'MAE: {mae:.4f}')
        print(f'R²: {r2:.4f}')
        
        return model_fit, forecast_series, {'rmse': rmse, 'mae': mae, 'r2': r2}
    except Exception as e:
        print(f"Error calculating metrics: {str(e)}")
        return model_fit, forecast_series, {'rmse': np.nan, 'mae': np.nan, 'r2': np.nan}

def find_best_ets_model(train, test):
    """
    Try different ETS configurations and return the best one
    """
    # Models to try
    ets_models = [
        ("Simple Exponential Smoothing", {"trend": None, "seasonal": None, "seasonal_periods": None}),
        ("Holt's Method", {"trend": "add", "seasonal": None, "seasonal_periods": None}),
        ("Holt-Winters Additive", {"trend": "add", "seasonal": "add", "seasonal_periods": 7}),
        ("Holt-Winters Multiplicative", {"trend": "add", "seasonal": "mul", "seasonal_periods": 7}),
        ("No Trend, Additive Seasonal", {"trend": None, "seasonal": "add", "seasonal_periods": 7}),
        ("No Trend, Multiplicative Seasonal", {"trend": None, "seasonal": "mul", "seasonal_periods": 7})
    ]
    
    best_model = None
    best_fit = None
    best_forecast = None
    best_metrics = {"rmse": float('inf')}
    
    for name, params in ets_models:
        print(f"\nTrying {name}...")
        try:
            # Create and fit model
            model = ExponentialSmoothing(
                train,
                trend=params["trend"],
                seasonal=params["seasonal"],
                seasonal_periods=params["seasonal_periods"],
                use_boxcox=False,
                initialization_method="estimated"
            )
            
            model_fit = model.fit()
            
            # Forecast
            forecast = model_fit.forecast(len(test))
            forecast_series = pd.Series(forecast, index=test.index)
            
            # Make sure both test and forecast don't have NaN values
            valid_idx = ~(np.isnan(test) | np.isnan(forecast))
            
            if sum(valid_idx) < 2:
                print("WARNING: Not enough valid data points for evaluation.")
                continue
            
            # Filter out NaN values
            test_clean = test[valid_idx]
            forecast_clean = forecast[valid_idx]
            
            # Calculate metrics
            rmse = np.sqrt(mean_squared_error(test_clean, forecast_clean))
            mae = mean_absolute_error(test_clean, forecast_clean)
            r2 = r2_score(test_clean, forecast_clean)
            
            print(f"{name} - RMSE: {rmse:.4f}, MAE: {mae:.4f}, R²: {r2:.4f}")
            
            # Keep track of best model
            if rmse < best_metrics["rmse"]:
                best_model = name
                best_fit = model_fit
                best_forecast = forecast_series
                best_metrics = {"rmse": rmse, "mae": mae, "r2": r2}
                best_params = params
        except Exception as e:
            print(f"Error fitting {name}: {str(e)}")
    
    # Plot best model results
    if best_fit is not None:
        plt.figure(figsize=(15, 6))
        plt.plot(train[-30:], label='Training Data (last 30 points)')
        plt.plot(test, label='Actual')
        plt.plot(best_forecast, label='Forecast')
        plt.title(f'Best ETS Model: {best_model} - Forecast vs Actual')
        plt.legend()
        plt.grid(True)
        plt.tight_layout()
        plt.show()
        
        print(f"\nBest ETS Model: {best_model}")
        print(f"Parameters: {best_params}")
        print(f"RMSE: {best_metrics['rmse']:.4f}")
        print(f"MAE: {best_metrics['mae']:.4f}")
        print(f"R²: {best_metrics['r2']:.4f}")
    
    return best_model, best_fit, best_forecast, best_metrics

def analyze_water_level_station_ets(file_path, station_name):
    """
    Process, analyze and build ETS model for water level station data
    """
    print(f"\n{'='*80}")
    print(f"Analyzing station: {station_name}")
    print(f"{'='*80}")
    
    # Read data
    try:
        df = pd.read_csv(file_path)
    except Exception as e:
        print(f"Error reading file {file_path}: {str(e)}")
        return None
    
    # Check if data exists
    if df.empty:
        print(f"No data found for {station_name}")
        return None
    
    # Parse dates and set as index
    try:
        df.columns = ['datetime', 'unit', 'water_level', 'label']
        df['datetime'] = pd.to_datetime(df['datetime'])
        df.set_index('datetime', inplace=True)
    except Exception as e:
        print(f"Error parsing dates: {str(e)}")
        print("Columns in file:", df.columns)
        return None
    
    # Display basic info
    print("\nData Overview:")
    print(f"Time Range: {df.index.min()} to {df.index.max()}")
    print(f"Number of records: {len(df)}")
    print(f"Number of missing values: {df['water_level'].isna().sum()}")
    
    # Handle missing values if any
    if df['water_level'].isna().any():
        df['water_level'] = df['water_level'].interpolate(method='time')
    
    # Plot water level data
    plt.figure(figsize=(15, 6))
    plt.plot(df['water_level'])
    plt.title(f'Water Level - {station_name}')
    plt.xlabel('Date')
    plt.ylabel('Water Level (m)')
    plt.grid(True)
    plt.tight_layout()
    plt.show()
    
    # Resample to daily frequency (using mean) to reduce noise
    daily_data = df['water_level'].resample('D').mean()
    
    # Handle any remaining NaN values in daily data
    if daily_data.isna().any():
        print(f"Filling {daily_data.isna().sum()} NaN values in daily data")
        daily_data = daily_data.interpolate(method='time')
    
    # Split data into train and test sets
    train_size = int(len(daily_data) * 0.8)
    train, test = daily_data[:train_size], daily_data[train_size:]
    
    print(f"\nTraining set size: {len(train)}")
    print(f"Test set size: {len(test)}")
    
    # Find best ETS model
    print("\nFinding best ETS model...")
    best_model, best_fit, best_forecast, best_metrics = find_best_ets_model(train, test)
    
    return {
        'station_name': station_name,
        'model_name': best_model,
        'model': best_fit,
        'forecast': best_forecast,
        'metrics': best_metrics
    }

# Main execution
def main():
    base_dir = r"c:\Users\RushenSamodya\Documents\GitHub\Final_Year_Project\labeled_data"
    
    # Path to Kelani River water level data
    kelani_dir = os.path.join(base_dir, "water_level_data", "kelani_ganga")
    
    # Check if directory exists
    if not os.path.exists(kelani_dir):
        print(f"Directory does not exist: {kelani_dir}")
        return
    
    # Get all stations for Kelani River
    stations = [f for f in os.listdir(kelani_dir) if f.endswith('.csv')]
    
    if not stations:
        print(f"No station data found in {kelani_dir}")
        return
    
    print(f"Found {len(stations)} stations for Kelani River")
    
    # Results container
    results = []
    
    # Process each station
    for station_file in stations:
        station_path = os.path.join(kelani_dir, station_file)
        station_name = station_file.replace('.csv', '')
        
        # Analyze station data
        result = analyze_water_level_station_ets(station_path, station_name)
        
        if result:
            results.append(result)
    
    # Compare results across stations
    if results:
        print("\n\nComparison of model performance across stations:")
        comparison = pd.DataFrame([
            {
                'Station': r['station_name'],
                'Best ETS Model': r['model_name'],
                'RMSE': r['metrics']['rmse'],
                'MAE': r['metrics']['mae'],
                'R²': r['metrics']['r2']
            }
            for r in results
        ])
        
        print(comparison)
        
        # Save results to CSV
        comparison.to_csv('ets_model_results.csv', index=False)
        print("Results saved to ets_model_results.csv")

# Execute main function
if __name__ == "__main__":
    main()
```
